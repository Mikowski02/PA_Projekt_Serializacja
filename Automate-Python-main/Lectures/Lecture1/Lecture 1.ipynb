## Dynamic types
Variables have a type that defines them. In Python, variables have no explicit type at declaration, just a value. The type is implicitly deduced. This is different in comparison to what you might have seen in **C** where variable types had to be strictly provided and the correct type for a variable was checked during compilation leading to warnings and errors. (Function identifiers in **C** are based on argument type!). In Python variables are **dynamically typed** as opposed to the **static typing** of **C**. This means that type is decided on run-time based on the variable value. This has benefits and allows for quick changes to be made to the code without the need to recompile the program. But at the same time might lead to errors that might have been avoided if type was strictly defined.

Comment on scope and name resolution:

In **C** we dealt with name resolution ruled by the use of `{}`. It was strict, in the sense that the variable was only available within the scope it was declared in, and not outside.

In Python, this is more relaxed. Python uses LEGB resolution (Local, Enclosing, Global, Built-in) instead of block scoping with `{}`. Variables in `if`, `for`, or `while` are visible outside (!), unlike **C** where scope ends at `}`. In the case of function, this is different.

```python
x = 1  # Global
def outer():
    x = 2  # Enclosing  
    def inner():
        x = 3  # Local
        print(x)  # 3 (local first)
    inner()
    print(x)  # 2 (enclosing)
outer()
print(x)  # 1 (global)
```

### More than one condition
Logical values are slightly different in Python than in **C**. Logical AND is realized with \& and OR with \|: For example:

```python
a = 10
```

```python
if a>-1 and a!=0:
    print(a)
```

```python
if (a>-1) & (a!=0): # note the (), it does not work without
    print(a)
```

```python
if a<-1 | a!=10:
    print(a)
```

```python
if a>-1 | a!=10:
    print(a)
```

```python
if a>-1 or a!=10:
    print(a)
```

Will not work exactly as you expect. Logical statements need to be additionally enclosed in ()-brackets, like this:

```python
if (a>-1) & (a!=0):
    print(a)
```

```python
if (a>-1) | (a!=10):
    print(a)
```

**Note:**, you can also perform logical operations with `and` and `or`!:

```python
if a<-1 or a!=0 and a > 0:
    print(a)
```

To circumvent this problem Python uses **pass** statement. For example:

```python
a = -10
if a>0:
    pass
else:
    print("Shell I pass?")
```

## Tuples - Krotka (?)
A bit different to a `list` type of collection is a `tuple`. Anything can be put into a tuple. Tuples can not be modified once created. To make one use `()`-brackets. A `tuple` holds an ordered set of data, of possibly different types (as lists do).

```python
a = 'ala'
b = 5.0
t = (a,b)
print(t)
```

Elements of a tuple are accessed with `[,]`-brackets:

```python
print(t[0])
print(t[1])
```

And modification attempts will fail:

```python
t[1] = 5.0
```

We might find tuples a convenient way to return more than one result from functions. Something that in **C** required passing values through pointers. For example:

```python
def fun():
    '''
    In this function a lot happens,
    things get calculated and in the end I need to return
    a list, an average and something else
    '''
    a = [ i**i for i in range(0,10)]
    s = sum(a) # a sum!
    # cast to float, just in case
    avg = float(s) / float(len(a));
    
    return a, ("s=", s), avg #!! a tuple in the return
```

```python
fun()
```

```python
res = fun()
print(type(res))
print(res)
for r in res:
    print(r)
```

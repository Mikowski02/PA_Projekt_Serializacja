# Task 0
Declare a simple class that represents a `Vehicle`. The class should have an `__init__` function which sets
* name
* max_speed
* mileage

```python

class Vehicle:
    color = 'White'
    def __init__(self, name='', max_speed=0, mileage=0):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage
    def print(self):
        print(f"Color: {self.color}, Name: {self.name}, Speed: {self.max_speed}, Age: {self.mileage}")

    def seating_capacity(self, capacity):
        return f"The seating capacity of a {self.name} is {capacity}."
    
    def fare(self, capacity):
        return f"Fare cost: {capacity*100}"
        
car1 = Vehicle('Volvo', 180, 12)
car1.print()
```

## Task 2
Add a class `Bus` and a property `capacity` and add `.seating_capacity()` to the `Vehicle` method. For `Bus` the default value should be 50.

```python
class Bus(Vehicle):
    def __init__(self, name='', max_speed=0, mileage=0, capacity=50):
        super().__init__(name, max_speed, mileage)
        self.capacity = capacity

    def seating_capacity(self):
        return super().seating_capacity(self.capacity)

    def fare(self):
        base_fare = self.capacity * 100
        return f"Fare cost of bus: {base_fare * 1.1}"

car3 = Bus("Ogorek", 100, 10, 50)
car3.print()
print(car3.seating_capacity())
print(car3.fare())
```

## Task 5
Implement a class case. The class should store a creation date and four values E1, E2, E3 and E4. Write a function (not a method) to randomly generate dates and another one to generate values E1 - E4. The second function should return the values as tuple. Choose ranges for the values.

Use:

import random

Create a number of instances of case. Use filter and sort for to select values fulfilling a certain condition form and sort them according to creation date. 

```python
import random
import datetime

class case:
    def __init__(self, creation_date, E1, E2, E3, E4):
        self.creation_date = creation_date
        self.E1 = E1
        self.E2 = E2
        self.E3 = E3
        self.E4 = E4

    def __repr__(self):  # funkcja do poprawnego print()
        return f"Case(creation_date= {self.creation_date}, E1= {self.E1}, E2= {self.E2}, E3= {self.E3}, E4= {self.E4})"

    def print(self):
        print(f"Creation date: {self.creation_date}, E(1,2,3,4): {self.E1} , {self.E2} , {self.E3} , {self.E4}")

def generate_date(): #1
    year = random.randint(1900, 2100)
    month = random.randint(1, 12)
    day = random.randint(1, 28)
    return datetime.date(year, month, day)

def generate_value(): #2
    E1 = random.randint(0, 99)
    E2 = random.randint(0, 99)
    E3 = random.randint(0, 99)
    E4 = random.randint(0, 99)
    return (E1, E2, E3, E4)

# generowanie nieposortowanych list
cases = [case(generate_date(), *generate_value()) for _ in range(10)]

# filtrowanie od warunku
selected_cases = list(filter(lambda c: c.E1 > 9, cases))

# sortowanie listy list
selected_cases.sort(key=lambda c: c.creation_date)

print("Selected and sorted cases:")
for case in selected_cases:
    print(case) 
```
